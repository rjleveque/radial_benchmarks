For the problem described at
https://faculty.washington.edu/rjl/radial_benchmarks/test1/.

See that page for some 2D results near the bottom.

The `input_data` directory contains files used for the GeoClaw simulation.

This notebook shows the extent of these files and also converts the .asc raster files into netCDF, using new GeoClaw tools that are currently under development, and that create much smaller files.

```python
%matplotlib inline
```

```python
from pylab import *
from clawpack.geoclaw import topotools, dtopotools
from clawpack.visclaw import plottools
```

```python
datadir = './input_data'
```

## Read in files that are in ASCII raster format:

```python
topo15s = topotools.Topography(f'{datadir}/topo15s.asc', topo_type=3)
topo13s = topotools.Topography(f'{datadir}/topo13s.asc', topo_type=3)
topo3s = topotools.Topography(f'{datadir}/topo3s.asc', topo_type=3)
```

```python
print(f'15" topo has shape {topo15s.Z.shape}')
print(f'3" topo has shape {topo3s.Z.shape}')
print(f'1/3" topo has shape {topo13s.Z.shape}')
```

### 15 arcsecond topo over computational domain

```python
topo15s.plot(limits=(-3000,100));

x1_3s, y1_3s, x2_3s, y2_3s = (68, 56, 72, 59)
plottools.plotbox([x1_3s,x2_3s,y1_3s,y2_3s])
title('15" topo and box showing 3" region');
```

### 3 arcsecond topo over part of continental shelf

```python
topo3s.plot(limits=(-3000,20));
x1_13s = 71.2
x2_13s = 71.45
y1_13s = 58
y2_13s = 58.15
plottools.plotbox([x1_13s,x2_13s,y1_13s,y2_13s])
title('3" topo and box showing 1/3" region');
```

### 1/3 arcsecond topo over part of shore

```python
topo13s.plot(limits=(-30,10));

xtrans,ytrans = loadtxt(f'{datadir}/xy_transect.txt', unpack=True)
plot(xtrans, ytrans, 'k')
title('1/3" topo near shore and transect for comparing solution');
```

## Write out as netCDF files

```python
fname = f'{datadir}/topo15s.nc'
topo15s.write(fname, topo_type=4, compression=True)
fname = f'{datadir}/topo3s.nc'
topo3s.write(fname, topo_type=4, compression=True)
fname = f'{datadir}/topo13s.nc'
topo13s.write(fname, topo_type=4, compression=True)
```

```python
ls -lh input_data/topo*
```

## dtopo file (seafloor deformation dz)

```python
dtopo = dtopotools.DTopography(f'{datadir}/gaussian5m_dtopo_15s.dtt3', dtopo_type=3)
print(f'dtopo file has shape {dtopo.dZ.shape} (ntimes x nlat x nlon)')
```

```python
dtopo.plot_dZ_colors(t=1, dZ_interval=100);
```

### write as netcdf

```python
fname = f'{datadir}/gaussian5m_dtopo_15s.nc'
dtopo.write(fname, dtopo_type=4, compression=True)
```
